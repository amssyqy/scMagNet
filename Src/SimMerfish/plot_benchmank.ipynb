```r
library(ggplot2)
library(dplyr)
library(egg)

# Read data
data0 <- read.table('/data/qyyuan/project/ST_CP/JE/NC_review/Methods/Benchmark/plot/data_mapping_stereo-seq.txt', sep='\t', header=TRUE, row.names=1, check.names=FALSE)

# Method names and their paired Ours columns
# Pattern: col pairs (method, ours) in order
col_names <- colnames(data0)
# Method columns at odd positions (1, 3, 5, 7, 9 in 1-indexed after row labels)
# Actually let me check: CytoSPACE, Ours0, CellTrek, Ours1, CMAP, Ours2, SIMO, Ours3, SpaTrio, Ours4
# So indices 1,3,5,7,9 = methods; 2,4,6,8,10 = ours

method_idx <- seq(1, ncol(data0), by=2)
ours_idx <- seq(2, ncol(data0), by=2)
method_names <- col_names[method_idx]

# Metrics to loop over
metrics <- c("precision", "recall", "f1", "jaccard")

for (metric in metrics) {
  # Get values
  other_vals <- as.numeric(data0[metric, method_idx])
  ours_vals <- as.numeric(data0[metric, ours_idx])
  
  # Predicted pairs from TP + FP
  tp <- as.numeric(data0["TP", method_idx])
  fp <- as.numeric(data0["FP", method_idx])
  pred_pairs <- tp + fp
  
  df <- data.frame(
    other_method = method_names,
    other_value = other_vals,
    ours_value = ours_vals,
    Predicted_pairs = pred_pairs,
    stringsAsFactors = FALSE
  )
  
  # Sort by Predicted_pairs
  df$other_method <- factor(df$other_method, levels = df$other_method[order(df$Predicted_pairs)])
  
  # Long format
  long_df <- bind_rows(
    data.frame(x=df$other_method, method=df$other_method, value=df$other_value),
    data.frame(x=df$other_method, method="scMagNet", value=df$ours_value)
  )
  long_df$x <- factor(long_df$x, levels=levels(df$other_method))
  
  # Color palette
  colors <- c(
    "CytoSPACE" = "#9DB4E8",
    "CellTrek"  = "#e377c2",
    "CMAP"      = "#FFDAC1",
    "SIMO"      = "#B0E0E6",
    "SpaTrio"   = "#D4E09B",
    "scMagNet"  = "#E78523"
  )
  # Add names for methods not in the fixed list
  # If method names differ, assign automatically
  
  p <- ggplot(long_df, aes(x=x, y=value, color=method)) +
    geom_linerange(aes(ymin=0, ymax=value, group=method),
                   color="lightgray", size=0.4, position=position_dodge(0.8)) +
    geom_point(position=position_dodge(0.8), size=2) +
    labs(x="Comparison (sorted by # Predicted pairs)", y=metric, color="Method") +
    theme_article() +
    theme(axis.text.x=element_text(angle=45, hjust=1), text=element_text(size=9)) +
    scale_color_manual(values=colors)
  
  pdf(paste0(metric, "_lollipop.pdf"), height=2, width=3)
  print(p)
  dev.off()
}
```

```r
library(ggplot2)
library(dplyr)
library(egg)

data0 <- read.table('/data/qyyuan/project/ST_CP/JE/NC_review/Methods/Benchmark/plot/data_mapping_stereo-seq.txt',
                    sep='\t', header=TRUE, row.names=1, check.names=FALSE)

col_names  <- colnames(data0)
method_idx <- seq(1, ncol(data0), by=2)
ours_idx   <- seq(2, ncol(data0), by=2)
method_names <- col_names[method_idx]

# ---- Metric mapping: internal name -> display name ----
metric_map <- c(
  "precision" = "Precision",
  "recall"    = "Recall",
  "f1"        = "F1 score",
  "jaccard"   = "Jaccard similarity"
)

colors <- c(
  "CytoSPACE" = "#9DB4E8",
  "CellTrek"  = "#e377c2",
  "CMAP"      = "#FFDAC1",
  "SIMO"      = "#B0E0E6",
  "SpaTrio"   = "#D4E09B",
  "scMagNet"  = "#E78523"
)

for (metric in names(metric_map)) {
  
  metric_label <- metric_map[[metric]]   # display name
  
  other_vals <- as.numeric(data0[metric, method_idx])
  ours_vals  <- as.numeric(data0[metric, ours_idx])
  tp <- as.numeric(data0["TP", method_idx])
  fp <- as.numeric(data0["FP", method_idx])
  pred_pairs <- tp + fp
  
  df <- data.frame(
    other_method    = method_names,
    other_value     = other_vals,
    ours_value      = ours_vals,
    Predicted_pairs = pred_pairs,
    stringsAsFactors = FALSE
  )
  df$other_method <- factor(df$other_method,
                            levels = df$other_method[order(df$Predicted_pairs)])
  
  # ---- Build paired data ----
  left_df <- data.frame(x = 1, y = df$ours_value,
                        method = "scMagNet",
                        pair = as.character(df$other_method),
                        stringsAsFactors = FALSE)
  right_df <- data.frame(x = 2, y = df$other_value,
                         method = as.character(df$other_method),
                         pair = as.character(df$other_method),
                         stringsAsFactors = FALSE)
  seg_df <- data.frame(x = 1, xend = 2,
                       y = df$ours_value, yend = df$other_value,
                       pair = as.character(df$other_method),
                       method = as.character(df$other_method),
                       stringsAsFactors = FALSE)
  point_df <- bind_rows(left_df, right_df)
  
  # ---- Plot ----
  p <- ggplot() +
    geom_segment(data = seg_df,
                 aes(x = x, xend = xend, y = y, yend = yend, color = 'gray'),
                 size = 0.4, alpha = 0.7) +
    geom_point(data = point_df,
               aes(x = x, y = y, color = method),
               size = 2.5) +
    scale_x_continuous(
      breaks = c(1, 2),
      labels = c("scMagNet", "Other methods"),
      limits = c(0.7, 2.3)
    ) +
    labs(x = NULL, y = metric_label, color = "Method") +   # <-- display name here
    theme_article() +
    theme(
      text            = element_text(size = 9),
      legend.position = "right",
      axis.text.x     = element_text(angle = 45, hjust = 1)
    ) +
    scale_color_manual(values = colors)
  
  pdf(paste0('stereo-seq/',metric, "_paired.pdf"), height = 1.8, width = 2.5)
  print(p)
  dev.off()
  
  cat(sprintf("Saved: %s_paired.pdf (y label: %s)\n", metric, metric_label))
}
```
